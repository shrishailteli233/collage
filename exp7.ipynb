```python
import pandas as pd
```

```python
df=pd.read_csv("auto_mpg.csv")
```

```python
df["city_milage"]=df["mpg"]*0.75
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
mpg_list=[df['mpg'],df['city_milage']]
```

```python
plt.boxplot(mpg_list)
```

```python
plt.title('mpg vs city milage')
```

```python
plt.title('mpg vs city milage')
plt.ylabel('mpg')

```

```python
plt.xlabel([1,2],['average mpg','city mpg'])
```

```python
plt.show
```

```python
import seaborn as sns
```

```python
sns.scatterplot(data=df,x='horsepower',y='mpg')
plt.title('correlation between horsepower and mpg')
plt.xlabel('horsepower')
plt.ylabel('mpg')
plt.show()
correlation=df['horsepower'].corr(df['mpg'])
print(f"correlation between horsepower and mpg:{correlation:.3f}")
```
